The following code is based off of the BLS sample code. I edited a few things, but it is largely the same. The following series IDs correspond to my three chosen industries:


* CEU5552000001 = Finance and Insurance Industry
* CEU7071000001 = Arts, Entertainment, and Recreation Inustry
* CEU6562000001 = Health Care and Social Assistance Industry

The last two digits refer to the data type. I chose code 01 which corresponds to "ALL EMPLOYEES, THOUSANDS".


```python

import requests
import json
import prettytable
```

```python

#Grabbing data from 2006-2015
headers = {'Content-type': 'application/json'}
data = json.dumps({"seriesid": ['CEU5552000001','CEU7071000001', 'CEU6562000001'],"startyear":"2006", "endyear":"2015"})
p = requests.post('https://api.bls.gov/publicAPI/v2/timeseries/data/', data=data, headers=headers)
json_data = json.loads(p.text)
print (json_data)
#Omitted data that created table

# Limited at 10 years so running again with different set of dates
```

```python
# Code sample from the BLS website that builds table and file
for series in json_data2['Results']['series']:
    x=prettytable.PrettyTable(["series id","year","period","value","footnotes"])
    seriesId = series['seriesID']
    for item in series['data']:
        year = item['year']
        period = item['period']
        value = item['value']
        footnotes=""
        for footnote in item['footnotes']:
            if footnote:
                footnotes = footnotes + footnote['text'] + ','
        if 'M01' <= period <= 'M12':
            x.add_row([seriesId,year,period,value,footnotes[0:-1]])
    output = open('BLS_Data.txt','a')
    output.write (x.get_string())
    output.close()
```
